```python
def get_season_match_links(url):
    data = requests.get(url)
    soup = BeautifulSoup(data.text, 'html.parser')
    links = soup.select('table.stats_table')[0].find_all('a')
    links = [l.get("href") for l in links]
    links = list(set([l for l in links if '/matches/' in l and len(l)>22]))
    return links
```

```python
def scrape_season_match_data(url, dest):
    links=get_season_match_links(url)
    pd.DataFrame(list(map(lambda l: l.split('/')[-1], links))).to_csv(f'{dest}/all_matches.csv')
    for link in links:
        dfs=pd.read_html(f"https://fbref.com{link}")
        results=process_match_data(dfs)

        results[0].to_csv(f'{dest}/{link.split('/')[-1]}_match_stat.csv')
        results[1].to_csv(f'{dest}/{link.split('/')[-1]}_home_player_stat.csv')
        results[2].to_csv(f'{dest}/{link.split('/')[-1]}_away_player_stat.csv')
        results[3].to_csv(f'{dest}/{link.split('/')[-1]}_shot_stat.csv')
        time.sleep(15)
```
